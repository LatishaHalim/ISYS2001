## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
I am designing a simple expense classifier, help me write a code and program in python using (if/elif/else) that automatically allocates the total budget given to a fixed percentage budget allocation on one expenses per category (Rent [45%], Groceries [25%], Entertainment [15%], Transport [10%], Mobile data [5%].

 Expectations:

- Input: Total budget (float), list of expenses (Each expense is a dictionary with two keys: ”category” (string) and ”amount’ (float). Example: {”category”: “rent”,”amount”: 500.0}”), however if there are difference in usage of uppercase and lower case such as rent/Rent should be accepted)

- Logic**:** for each expense — allocated amount = category% × total budget → if expense ≤ allocated → "still on budget" → if expense > allocated → "over budget"

- Output**:** per-expense feedback line [”Rent : still on budget (spent $450.00 of $450.00 allocated)”

 [#] If an expense category doesn’t match one of the five listed categories, print unknown category

[#] Round both allocated amount and expense amount to 2 decimal places using Python's ____round ___ function. By that, perform the comparison using the rounded values, not the raw input values.

—> For the output design:

 [Make the output look pretty with appropriate spacing after each line and also make the per-expense feedback (spent "amount" of "amount" allocated) line appear right under "still on budget"/"over budget”]

 Make the program detailed enough that a classmate could paste it cold and get the same working tool
```

````python
# ==========================================
# Expense Classifier
# ==========================================
# Categories:
# Rent            - 45%
# Groceries       - 25%
# Entertainment   - 15%
# Transport       - 10%
# Mobile Data     - 5%
#
# The program:
# 1. Takes the total budget.
# 2. Takes one expense for each category.
# 3. Automatically calculates the allocated budget.
# 4. Compares the rounded expense with the rounded allocation.
# 5. Displays whether the expense is still on budget or over budget.
# ==========================================

# ----- Input -----
total_budget = float(input("Enter your total budget: $"))

print("\nEnter your expenses:")

# List of expense dictionaries
expenses = []

# Number of expenses to enter
number_of_expenses = int(input("How many expenses would you like to enter? "))

for i in range(number_of_expenses):
    print(f"\nExpense #{i + 1}")

    category = input("Category: ")
    amount = float(input("Amount spent: $"))

    # Store each expense as a dictionary
    expenses.append({
        "category": category,
        "amount": amount
    })

print("\n" + "=" * 50)
print("           EXPENSE CLASSIFICATION")
print("=" * 50)

# ----- Process Each Expense -----
for expense in expenses:

    # Make category lowercase so Rent/rent/RENT all work
    category = expense["category"].strip().lower()

    # Round expense amount BEFORE comparison
    spent = round(expense["amount"], 2)

    # Determine allocation using if/elif/else
    if category == "rent":
        allocation = round(total_budget * 0.45, 2)
        display_name = "Rent"

    elif category == "groceries":
        allocation = round(total_budget * 0.25, 2)
        display_name = "Groceries"

    elif category == "entertainment":
        allocation = round(total_budget * 0.15, 2)
        display_name = "Entertainment"

    elif category == "transport":
        allocation = round(total_budget * 0.10, 2)
        display_name = "Transport"

    elif category == "mobile data":
        allocation = round(total_budget * 0.05, 2)
        display_name = "Mobile Data"

    else:
        print(f"\nCategory: {expense['category']}")
        print("Unknown category")
        print("-" * 50)
        continue

    # Compare the rounded values
    if spent <= allocation:
        status = "Still on budget"
    else:
        status = "Over budget"

    # ----- Output -----
    print(f"\n{display_name}")
    print(f"Status : {status}")
    print(f"Spent  : ${spent:.2f} of ${allocation:.2f} allocated")
    print("-" * 50)

print("\nExpense classification complete.")
````
