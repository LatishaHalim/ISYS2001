## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:
[Write your prompt here]
"""
# Ask the user for their monthly income
monthly_income = float(input("Enter your monthly income: $"))

# Ask the user for up to 3 expenses
expense1 = float(input("Enter Expense 1: $"))
expense2 = float(input("Enter Expense 2: $"))
expense3 = float(input("Enter Expense 3: $"))

# Calculate total expenses
total_expenses = expense1 + expense2 + expense3

# Calculate remaining budget
remaining_budget = monthly_income - total_expenses

# Display the results
print("\n----- Budget Summary -----")
print(f"Monthly Income: ${monthly_income:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")

# Warn the user if the budget is negative
if remaining_budget < 0:
    print("Warning: Your budget is negative! You are spending more than you earn.")
else:
    print("Good job! Your budget is within your income.")
```

```python
# ENHANCED BUDGET TRACKER
# [Paste your enhanced code here]
# ==========================================
# Enhanced Monthly Budget Tracker
# ==========================================

print("====== Monthly Budget Tracker ======")

# Get user's name
name = input("Enter your name: ")

# Get monthly income
monthly_income = float(input("Enter your monthly income ($): "))

# Get expenses by category
print("\nEnter your monthly expenses:")

food = float(input("🍔 Food Expense ($): "))
transport = float(input("🚌 Transport Expense ($): "))
entertainment = float(input("🎮 Entertainment Expense ($): "))

# Calculate totals
total_expenses = food + transport + entertainment
remaining_budget = monthly_income - total_expenses

# Display summary
print("\n========== Budget Summary ==========")
print(f"Hello, {name}!")
print(f"Monthly Income      : ${monthly_income:.2f}")
print(f"Food Expense        : ${food:.2f}")
print(f"Transport Expense   : ${transport:.2f}")
print(f"Entertainment Expense: ${entertainment:.2f}")
print("------------------------------------")
print(f"Total Expenses      : ${total_expenses:.2f}")
print(f"Remaining Budget    : ${remaining_budget:.2f}")

# Find the largest expense
largest_expense = max(food, transport, entertainment)

if largest_expense == food:
    print("💡 Your highest spending category is Food.")
elif largest_expense == transport:
    print("💡 Your highest spending category is Transport.")
else:
    print("💡 Your highest spending category is Entertainment.")

# Budget status
if remaining_budget < 0:
    print("\n⚠️ Warning! You are over your budget.")
    print("Try reducing your expenses next month.")
elif remaining_budget < monthly_income * 0.2:
    print("\n🟡 You have less than 20% of your income remaining.")
    print("Consider saving more where possible.")
else:
    print("\n🟢 Excellent! You are managing your budget well.")

print("====================================")
```
