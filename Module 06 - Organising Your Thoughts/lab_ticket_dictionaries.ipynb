## The three functions your code has to get right

Build your dictionary however you like, then write these three functions to work on it.
The checks at the bottom call them by name, so use these names and have them behave as
described:

- `total_for(spending, category)` returns the amount spent in that category, and returns

- `add_transaction(spending, transaction)` takes one transaction, in the same shape as

```python
# PLAN:
#     Organise these transactions
#     Choose their key and the value they hold (Key: Category, Value: Amount)
#     Write 3 functions that work for the dictionary
#     ADD a new dictionary (BUDGET)
#     Compare budget and transaction and print whether the result is "OVERBUDGET" or "WITHIN BUDGET"


# Build your dictionary from the transactions, and write your three functions here.

# Build a simplified dictionary
spending = {'rent': 1200, 'groceries': 165, 'transport':90, 'eating out': 52, 'phone': 40, 'entertainment':40, 'textbook': 95}

# If the box doesn't exist (a safe way to look up something that doesn't exist)
spending.get("gym", "None listed")

# Total_For (FUNCTION 1)
def total_for(spending, category):  # create a function
  return spending.get(category, 0) # will return it as 0 not error

print(total_for(spending, 'rent'))
print(total_for(spending,'gym'))

# add_transaction (FUNCTION 2)
# spending and transaction is the parameter
def add_transaction(spending, transaction):
  category = transaction['category']
  amount = transaction['amount']
  if category in spending:  # Checks whether the key already exists, if it does you add new amount to the existing total, else..
    spending[category] = spending[category] + amount
  else:
    spending[category] = amount

add_transaction(spending, {'date': '2026-08-30', 'category': 'groceries', 'amount': 20, 'description': 'Snacks'})
print(spending)

# summary(spending) (FUNCTION 3)
def summary(spending):
  result = {}
  for category, amount in spending.items():
    result[category] = amount
  return result

print(summary(spending))
```

```python
# Your own feature goes here (BUDGET COMPARISON)

# Spending dictionary
spending = {'rent': 1200,
            'groceries': 165,
            'transport':90,
            'eating out': 52,
            'phone': 40,
            'entertainment': 40,
            'textbook': 95
            }

# New budget dictionary
budget ={
    'rent': 1000,
    'groceries': 300,
    'transport': 150,
    'eating out': 100,
    'phone': 80,
    'entertainment': 50,
    'textbook': 100
}
for category in budget:
  actual = spending.get(category, 0) # Use .get() for safe retrieval, default to 0 if not found
  allowed = budget[category]
  difference = allowed - actual

  if difference < 0:
   print (f"{category}: {abs(difference)} = OVERBUDGET")
  else:
    print (f"{category}: {difference} = WITHIN BUDGET")
```
