```python
# Import pandas as pd

# Read the whole table in one line.
import pandas as pd

df = pd.read_csv('sales_data.csv')

df
```

```python
# To confirm data types  before cleaning
df.dtypes
# If showing no numbers, it means that they're showing up as strings in pandas
```

```python
# Clean the amounts in the column once, for every row at the same time
# Select the column first through df [] and turn it into numeric
df['Quantity'] = df['Quantity'].str.replace('$', '', regex=False)
df['Quantity'] = pd.to_numeric(df['Quantity'])

df['Unit_Price'] = df['Unit_Price'].str.replace('$', '', regex=False)
df['Unit_Price'] = pd.to_numeric(df['Unit_Price'])

df['Total_Sale'] = df['Total_Sale'].str.replace('$', '', regex=False)
df['Total_Sale'] = pd.to_numeric(df['Total_Sale'])


df
```

```python
# To confirm data types: Quantity as (int64), Unit_Price as (float64), Total_Sale as (float64)
df.dtypes
```

```python
# Task 1: total sales by region, highest first.
region_totals_pandas = df.groupby('Region')['Total_Sale'].sum()   # Group them per regions and total it
region_totals_pandas = region_totals_pandas.sort_values(ascending=False)
# Sort them

region_totals_pandas
```

```python
# Task 2: total units sold per product.
product_totals_pandas = df.groupby('Product')['Quantity'].sum()   # Group them by product and total the quantity sold
product_totals_pandas = product_totals_pandas.sort_values(ascending=False)

product_totals_pandas

```
